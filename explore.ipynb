```python
import pandas as pd
import plotly.express as px

rows = []
with open("nino34.txt") as f:
    for line in f:
        parts = line.split()
        if len(parts) == 13 and parts[0].isdigit():
            year = int(parts[0])
            for month, val in enumerate(parts[1:], start=1):
                v = float(val)
                if v > -90:  # skip missing values
                    rows.append((year, month, v))

df = pd.DataFrame(rows, columns=["year", "month", "anomaly"])
df["date"] = pd.to_datetime(dict(year=df.year, month=df.month, day=1))

print(df.head())
print(df.tail())
```

```python
fig = px.line(df[df.year >= 1990], x="date", y="anomaly",
              title="Niño 3.4 Anomaly (1990–present)")
fig.add_hline(y=0.5, line_dash="dash", line_color="orange",
              annotation_text="El Niño threshold")
fig.add_hline(y=2.0, line_dash="dash", line_color="red",
              annotation_text="Very strong")
fig.show()
```
